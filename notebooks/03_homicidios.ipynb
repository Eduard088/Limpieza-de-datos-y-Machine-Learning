## Las 10 Principales Motivos de Muerte en México:

```python
(
    homicidios['causa_def']
    .value_counts()
    .sort(by = 'count', descending= True)
    .head(10)
)
```

## Muertes por Tipo de Defunción en México:

```python
(
    homicidios['tipo_defun']
    .value_counts()
    .sort(by = 'count', descending= True)
)
```

## Muertes por Tipo de Defunción y Sexo en México:

```python
(
    homicidios
    .group_by(
        ['sexo', 'tipo_defun']
    )
    .len()
    .rename(
        {'len': 'Frecuencia'}
    )
    .sort(by = 'Frecuencia', descending= True)
    .head(10)
)
```

## Muertes de Mujeres por Tipo de Defunción en México:

```python
(
    homicidios
    .filter(pl.col('sexo') == 'Mujer')
    .group_by(
        ['sexo', 'tipo_defun']
    )
    .len()
    .rename(
        {'len': 'Frecuencia'}
    )
    .sort(by='Frecuencia', descending= True)
)
```

## Principales Sitios donde Ocurrieron Homicidios de Mujeres:

```python
(
homicidios
    .filter(
    (pl.col("tipo_defun") == "Homicidio (Agresión)") &
    (pl.col("sexo") == "Mujer")
    )
    .group_by(
        ['sitio_ocur', 'tipo_defun']
    )
    .len()
    .rename(
        {'len': 'Frecuencia'}
    )
    .sort(by = 'Frecuencia', descending= True)
)
```

## Muertes de Mujeres por Tipo de Defunción  y Sitio donde Ocurrieron en México:

```python
(
    homicidios
    .filter(pl.col('sexo') == 'Mujer')
    .group_by([
        'tipo_defun', 'sitio_ocur'
    ]
    )
    .len()
    .rename(
        {'len': 'Frecuencia'}
    )
    .sort(by = 'Frecuencia', descending= True)
)
```

## Principales Lugares donde Ocurrieron Homicidios en México:

```python
(
    homicidios
    .filter(pl.col("tipo_defun") == "Homicidio (Agresión)")
    .group_by([
        "sitio_ocur", "tipo_defun"
    ])
    .len()
    .rename({'len': 'Frecuencia'})
    .sort('Frecuencia', descending = True)
    .head(10)
)
```
