## Método 3: Calcular valores de p manualmente usando Wald test:

```python
def calculate_p_values_manual(model_sklearn, X_train):
    """
    Calcula valores de p usando el test de Wald
    """
    # Obtener coeficientes y matriz de covarianza estimada
    coef = model_sklearn.coef_[0]
    
    # Calcular matriz de información de Fisher aproximada
    # Usar predicciones del modelo entrenado
    y_pred_proba = model_sklearn.predict_proba(X_train)[:, 1]
    
    # Weights para la matriz de información
    weights = y_pred_proba * (1 - y_pred_proba)
    
    # Matriz de información de Fisher
    X_weighted = X_train.values * weights.reshape(-1, 1)
    fisher_info = np.dot(X_train.values.T, X_weighted)
    
    # Intentar invertir la matriz (agregar regularización si es necesario)
    try:
        cov_matrix = np.linalg.inv(fisher_info)
    except np.linalg.LinAlgError:
        # Si la matriz es singular, agregar regularización
        cov_matrix = np.linalg.inv(fisher_info + np.eye(fisher_info.shape[0]) * 1e-8)
    
    # Errores estándar
    std_errors = np.sqrt(np.diag(cov_matrix))
    
    # Estadística de Wald
    wald_stats = coef / std_errors
    
    # Valores de p (distribución normal estándar, test de dos colas)
    p_values = 2 * (1 - chi2.cdf(wald_stats**2, df=1))
    
    return p_values, std_errors, wald_stats
```

## Calculamos Valores de p:

```python
p_values, std_errors, wald_stats = calculate_p_values_manual(model_sklearn, X_train)
```

## Creamos Dataframe con los Resultados:

```python
results_df = pd.DataFrame({
    'Variable': X_train.columns,
    'Coeficiente': model_sklearn.coef_[0],
    'Error_Estandar': std_errors,
    'Estadistica_Wald': wald_stats,
    'Valor_P': p_values,
    'Odds_Ratio': np.exp(model_sklearn.coef_[0])
})
```

## Validación Cruzada:

```python
print("\n===== VALIDACIÓN CRUZADA =====")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
accuracy_scores = []
balanced_acc_scores = []
f1_scores = []

for train_index, test_index in skf.split(X_resampled, y_resampled):
    X_train_fold, X_test_fold = X_resampled.iloc[train_index], X_resampled.iloc[test_index]
    y_train_fold, y_test_fold = y_resampled[train_index], y_resampled[test_index]
    
    model_sklearn.fit(X_train_fold, y_train_fold)
    y_pred_fold = model_sklearn.predict(X_test_fold)
    
    accuracy_scores.append(accuracy_score(y_test_fold, y_pred_fold))
    balanced_acc_scores.append(balanced_accuracy_score(y_test_fold, y_pred_fold))
    f1_scores.append(f1_score(y_test_fold, y_pred_fold))
```

```python
print(f"Accuracy promedio: {np.mean(accuracy_scores):.4f}")
print(f"Balanced Accuracy promedio: {np.mean(balanced_acc_scores):.4f}")
print(f"F1-score promedio: {np.mean(f1_scores):.4f}")
```

## Función para el Análisis de Variables Específicas:

```python
def analizar_variables_especificas(results_df, keywords=None):
    """
    Analiza variables específicas basadas en palabras clave
    """
    if keywords is None:
        keywords = ['edad', 'estado', 'municipio', 'escolaridad', 'ocupacion']
    
    print(f"\n===== ANÁLISIS DE VARIABLES ESPECÍFICAS =====")
    for keyword in keywords:
        matching_vars = results_df[results_df['Variable'].str.contains(keyword, case=False, na=False)]
        if len(matching_vars) > 0:
            print(f"\nVariables relacionadas con '{keyword}':")
            print(matching_vars[['Variable', 'Coeficiente', 'Valor_P', 'Odds_Ratio']].head(10).to_string(index=False))
```
